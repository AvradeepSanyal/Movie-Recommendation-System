TURNING COLUMNS INTO CLEAN AND RIGHT FORMAT

For GENRE

```python
movies.iloc[0].genres #accessing genres column from movies db
```

```python
import ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj): #ast.literal_eval() safely converts a string containing a Python literal (like a list, dictionary, number, tuple, string, boolean, or None) into its actual Python data structure
        L.append(i['name'])
    return L

#from dict-list to list in genres
```

```python
movies['genres'] = movies['genres'].apply(convert) #making it permanent and applying the func
```

For KEYWORD

```python
movies['keywords'] = movies['keywords'].apply(convert) #same steps for KEYWORD and applying the func
```

```python
movies.head(1) #checking if changes occured in keywords and genres
```

For CAST

```python
 def convert(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter += 1
        else:
            break
    return L

#as we wanted only top 3 actors from cast (as top 3 cast are at first we counted till 3)
```

```python
movies['cast'] = movies['cast'].apply(convert) #making it permanent and applying the func
```

```python
movies.head(1) #checking until now again if changes occured
```

For CREW

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L

#as we wanted just the director name from crew
```

```python
movies['crew'] = movies['crew'].apply(fetch_director) #making it permanent and applying the func
```

```python
movies.head(1) #checking if changes are permanent till now
```

For OVERVIEW

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split()) #as we wanted str to list (and making it permanent)
```

```python
movies.head(1) #checking again
```

NOW, All done! But now we will do a last transformation in our list in every column of our df, we will remove space from every words, AFTER making a new updated db we will - make it lowercase, change any formats, so our model dont treat it as 2 different entity but rather as 1 entity(as we are making tags) and it helps in vectorixation. (As we need to find similarity in our recommendation system based on tags thats why this!)

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

! COME HERE AFTER LINE 43 !

```python
import nltk

from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)

new_df['tags'] = new_df['tags'].apply(stem)
```

# 2. Vectorization

As we need to find similarity based on tags, we can only do that by doing finding the text similarity betweem 2 different rows/TAGS (basically here 2 diff movies as 1 row depicts all the info for a specific movie). We cant do that in text format we can only do that with number format. So we will convert this whole text to vector(set of numbers for each row/TAGS) and then we can do similarity matching between them. THATS WHAT WE GONNA DO HERE!

```python
new_df['tags'][0]
```

```python
new_df['tags'][1]
```

NOW, we will use the simplest technique BAG OF WORDS (Best approach - combined text/large text should always be minimum, as it results in less dimensionality and time)
1. Combine all tags or words
2. THEN from that large text(ex- 5000 words) we will pick up the most common words and make a table of columns with each word as each column
3. THEN calculate the frequency for each word against every row/movie and observe which word occurs multiple times for each movie
4. THEN the set of numbers we get for each row/movie is the vector of that movie
Then, plot [(movie, words)] all rows or movie in a N/words-dimensional space(as its a vector) 
Then if someone likes a movie/now set of vectors in n-dim space we will fecth/recommend the closest 5 vectors/movies,
Note - we wont include STOP WORDS (in, a, to...), (as they are used for senetence formation not sentence meaning)

LETS START!
We will use count vectorizer class

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray() 
#converting to a numpy array (as countvectorizer by default converts this into scipy sparse matrix but we need array here)
```

```python
vectors.shape #4806 movies/vectors of that movie, 5000 words
```

```python
vectors #every movie in vector format
```

```python
vectors[0] #first movie vector
```

NOW, we will do stemming as it makes all same words with slight variation to its default format -
[loved, loving] to [love, love]
we will do it in our new data (new_df), using nltk
- "Go to Line 36"

NOW, we have to calculate distance of each movie to every other movie (the more distance/degree = less similarity)
- In here we will calculate COSINE SIMILARITY (angle) not euclidean distance (distance) [more high dimensional more failure of euclidean distance]
- we will do this using a scikir-learn class -> cosine_similarity

```python
from  sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors) #storing all calculation of distances of every movie with another using cosine similarity
```

```python
similarity.shape #why same? as we are doing EVERY MOVIE DISTANCE CALCULATION WITH EVERY OTHER MOVIE
```

```python
similarity[0] #this array is the similarity/distance of first movie with every other movie (thats why first is 1)
```

```python
similarity[1] #here second is 1
```

# 3. Main Function

What we will do here ?
1. Make function to fetch for the index (raw step-by-step code in line-'49')
2. Using that index, we will access the same index from similarity ( " )
3. Then sort from the highest similarity to lowest (we cant do normal sorting as we lose our index position thats why we will use 'Enumerate' then sort from 2nd index)(raw step-by-step code in line-'50')
4. Atlast return or RECOMMEND the top 5 similar movies/vector

```python
#1 and 2
new_df[new_df['title'] == 'Batman Begins']
print(new_df[new_df['title'] == 'Batman Begins'].index) #DOUBT - how index[0] works to fecth the real index not only .index
```

```python
#3.
#first step
print(enumerate(similarity[0])) #DOUBT - wth! enumerate do here & ingeneral
#second step
print(list(enumerate(similarity[0]))) #distance of 0th movie/first movie with itself and others[sec, third, forth...]
#third step
print(sorted(list(enumerate(similarity[0])), reverse=True))#with reverse=true means descending as we need highhest to lowest similarity/top 5 movies similar to what we say
#fourth step in sorted after reverse is [ key=lambda x:x[1] what key means and x[1] means sort from second, second is first now thats why after its [1:6] not [0:6] right? <-- DOUBT
#fifth step - [1:6] means excluding (ex-'batman begins'), show first 5 movies. 6 is excluded it only counts till 5th index
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0] #1
    distances = similarity[movie_index] #2
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6] #3

    for i in movies_list:
        print(new_df.iloc[i[0]].title) #raw = new_df.iloc[index val].title
```

```python
recommend('Batman Begins') #4
```
